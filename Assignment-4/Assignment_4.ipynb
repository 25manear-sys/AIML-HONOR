```python
import pandas as pd, os
class EmpPandas:
    def __init__(self):
        self.file = "emp_pd.csv"
        if not os.path.exists(self.file):
            pd.DataFrame(columns=["ID", "Name", "Dept", "Salary"]).to_csv(self.file, index=False)
    def create(self):
        df = pd.read_csv(self.file)
        i = int(input("ID: "))
        n = input("Name: ")
        d = input("Dept: ")
        s = float(input("Salary: "))
        new_row = pd.DataFrame([{"ID": i, "Name": n, "Dept": d, "Salary": s}])
        df = pd.concat([df, new_row], ignore_index=True)
        df.to_csv(self.file, index=False)
        print("Added successfully")
    def read(self):
        df = pd.read_csv(self.file)
        print(df)
    def update(self):
        df = pd.read_csv(self.file)
        i = int(input("ID to update: "))
        if i in df["ID"].values:
            df.loc[df["ID"] == i, "Name"] = input("New Name: ")
            df.loc[df["ID"] == i, "Dept"] = input("New Dept: ")
            df.loc[df["ID"] == i, "Salary"] = float(input("New Salary: "))
            df.to_csv(self.file, index=False)
            print("Updated successfully")
        else: print("ID not found")
    def delete(self):
        df = pd.read_csv(self.file)
        i = int(input("ID to delete: "))
        if i in df["ID"].values:
            df = df[df["ID"] != i]
            df.to_csv(self.file, index=False)
            print("Deleted successfully")
        else: print("ID not found")
def main():
    e = EmpPandas()
    while True:
        c = input("\n1.Create 2.Read 3.Update 4.Delete 5.Exit\nChoice: ")
        if c == "1": e.create()
        elif c == "2": e.read()
        elif c == "3": e.update()
        elif c == "4": e.delete()
        elif c == "5": break
if __name__ == "__main__": main()
```
