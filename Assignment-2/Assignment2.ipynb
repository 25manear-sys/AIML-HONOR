```python
#Assignment 2
import csv, os
class EmpCSV:
    def __init__(self):
        self.file = "emp.csv"
        if not os.path.exists(self.file):
            with open(self.file, "w", newline="") as f:
                csv.writer(f).writerow(["ID", "Name", "Dept", "Salary"])
    def create(self):
        i = input("ID: ")
        n = input("Name: ")
        d = input("Dept: ")
        s = input("Salary: ")
        with open(self.file, "a", newline="") as f:
            csv.writer(f).writerow([i, n, d, s])
        print("Added successfully")
    def read(self):
        with open(self.file, "r") as f:
            for r in csv.reader(f): print(r)
    def update(self):
        i = input("ID to update: ")
        rows = []
        with open(self.file, "r") as f:
            reader = csv.reader(f)
            header = next(reader)
            rows.append(header)
            for r in reader:
                if r[0] == i:
                    r[1] = input("New Name: ")
                    r[2] = input("New Dept: ")
                    r[3] = input("New Salary: ")
                rows.append(r)
        with open(self.file, "w", newline="") as f:
            csv.writer(f).writerows(rows)
        print("Updated successfully")
    def delete(self):
        i = input("ID to delete: ")
        rows = []
        with open(self.file, "r") as f:
            reader = csv.reader(f)
            header = next(reader)
            rows.append(header)
            for r in reader:
                if r[0] != i: rows.append(r)
        with open(self.file, "w", newline="") as f:
            csv.writer(f).writerows(rows)
        print("Deleted successfully")
def main():
    e = EmpCSV()
    while True:
        c = input("\n1.Create 2.Read 3.Update 4.Delete 5.Exit\nChoice: ")
        if c == "1": e.create()
        elif c == "2": e.read()
        elif c == "3": e.update()
        elif c == "4": e.delete()
        elif c == "5": break
if __name__ == "__main__": main()
```
