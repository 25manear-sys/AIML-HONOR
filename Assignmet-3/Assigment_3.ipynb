```python
import sqlite3
class EmpDB:
    def __init__(self):
        self.conn = sqlite3.connect("emp.db")
        self.cur = self.conn.cursor()
        self.cur.execute("CREATE TABLE IF NOT EXISTS emp (id int, name text, dept text, sal real)")
        self.conn.commit()
    def create(self):
        i = int(input("ID: "))
        n = input("Name: ")
        d = input("Dept: ")
        s = float(input("Salary: "))
        self.cur.execute("INSERT INTO emp VALUES (?, ?, ?, ?)", (i, n, d, s))
        self.conn.commit()
        print("Added successfully")
    def read(self):
        self.cur.execute("SELECT * FROM emp")
        for r in self.cur.fetchall(): print(r)
    def update(self):
        i = int(input("ID to update: "))
        n = input("New Name: ")
        d = input("New Dept: ")
        s = float(input("New Salary: "))
        self.cur.execute("UPDATE emp SET name=?, dept=?, sal=? WHERE id=?", (n, d, s, i))
        self.conn.commit()
        print("Updated successfully")
    def delete(self):
        i = int(input("ID to delete: "))
        self.cur.execute("DELETE FROM emp WHERE id=?", (i,))
        self.conn.commit()
        print("Deleted successfully")
def main():
    e = EmpDB()
    while True:
        c = input("\n1.Create 2.Read 3.Update 4.Delete 5.Exit\nChoice: ")
        if c == "1": e.create()
        elif c == "2": e.read()
        elif c == "3": e.update()
        elif c == "4": e.delete()
        elif c == "5": break
if __name__ == "__main__": main()
```
